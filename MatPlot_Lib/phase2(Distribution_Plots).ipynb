# Distribution Plots
     These are used to visualise the Qualitative Dataset

 # Some Types are 
 1 Dist Plot / Hist plot 
 2 JOint PLot
 3 Pair Plot
 4 Rug Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=sns.load_dataset("tips")
data
```

```python
data['size'].unique()
```

```python
data['tip'].unique()
```

 # Distribution GRaph or Hist Graph

```python
sns.displot(data['total_bill'],kde=True)
```

```python
sns.histplot(data['tip'],bins=20,kde=True)
# kde is used to crweate a line showing all values throughout the graph
```

Creating Two Graps under a single canvas

```python
fig=plt.Figure()
axis1=fig.add_axes([0.1,0.1,0.5,0.2])

```

```python
plt.subplot(1,2,1)
axis2=sns.histplot(data["tip"],kde=True,color="Red")
plt.subplot(1,2,2)
sns.histplot(data["total_bill"],kde=True,color="Black")
```

 Joint Plot - It is used to comapre the two graph

```python
sns.jointplot(x=data["total_bill"],y=data["tip"],data=data,kind='scatter',color="Red")
```

```python
sns.jointplot(x=data["total_bill"],y=data["tip"],data=data,kind='kde',color="Red")
```

```python
sns.jointplot(x=data["total_bill"],y=data["tip"],data=data,kind='hex',color="Red")
```

```python
sns.jointplot(x=data["total_bill"],y=data["tip"],data=data,kind='reg',color="Red")
```

PairPlot

```python
sns.pairplot(data,hue="sex")
```

```python
sns.pairplot(data,hue="size",palette="rainbow")
```

RugPlot

```python
sns.rugplot(data)
```

```python
sns.rugplot(data['tip'])
```
