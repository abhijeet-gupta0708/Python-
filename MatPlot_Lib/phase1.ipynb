```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace(0,5,11)
x

```

```python
y=x**3
y
```

# PLOTTING

```python

plt.plot(x,y)
plt.title("FirstGraph")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

# Subplot

```python
plt.subplot(2,2,1)
plt.plot(y*2,x*3)
plt.subplot(2,2,2)
plt.plot(x,x)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.show
```

# Object Oriented Way

it is a way of controlling the canvas

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.1,0.1,0.3,0.3])
axis2.plot(x,y)
```

# Types of Plots

```python
fig=plt.figure(figsize=(6,6))
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.6,0.3,0.3])
axis2.plot(x,y)
```
