# Q 1 : Make a new column for episode count

```python
def episode_count(txt) :
    newdata =""
    check=False
    for i in txt :
        if i==")" :
            check=False
            break
        if check==True :
            newdata=newdata+i
        if i=="(" :
            check=True
    return newdata   
       
    
data["Episode Count"] = data["Title"].apply(episode_count)

        
```

```python
data.head()
```

```python
def episode_number (txt) :
    check=False
    newdata= ""
    for i in txt :
        if i=="e":
            check=True;
        if check == True :
           break
        if check== False :
            newdata=newdata+i
    return newdata   
```

Removing Eps from Episode COunt

```python
data["Episode Count"]=data["Episode Count"].apply(episode_number)
data.head()
```

```python
data["Episode Count"].dtype
```

```python
# COnverting the STring Type of Episode COunt to integer
data["Episode Count"]=data["Episode Count"].astype(int)
```

 # Q : 2 Make a new column for time stamp

```python
data.loc[5]["Title"]
```

```python
def start_time(txt) :
    check=False
    newdata=""
    for i in txt :
        if check==True :
            if i=="-" or i== "--" :
                check=False
                break   
            newdata=newdata+i
        if i==")":
            check=True
            continue
        
        
    return newdata

data["Start Time"]=data["Title"].apply(start_time)
data.head()
```

 Writing the End Time

```python
data.loc[3]["Title"]
```

```python
def end_time(txt) :
    check=False
    newdata=""
    for i in range (len(txt)) :
        if txt[i]=="-" :
            for j in range (10) :
                newdata=newdata+txt[j+i+1]
        
    return newdata

data["End Time"]=data["Title"].apply(end_time)
data.head()
```

As you can see that when we try to use this method some of the title might have another - , so we are not able to get our desired product
But there is one more observation that all the date start Time +End Time start just after episode number or  after ")", so we will count that

```python
data.loc[3]["Title"]
```

```python
def time_count (txt) :
    newdata =""
    for i in range (len (txt)) :
        if txt[i]==")" :
            for j in range (1,20):
                newdata=newdata+txt[j+i]
    
    return newdata

data["Time"]=data["Title"].apply(time_count)
data.head()
```

 So we got the Whole time length of our Anime Series , so now deleting the Columns We Created As " Start Time and End Time "

```python
# Deleting the Column of Start Time 
data.drop("Start Time",axis=1,inplace=True)
data.head()
```

```python
# Deleting the Column of End Time
data.drop("End Time",axis=1,inplace=True)
data.head()
```

# Q 7 :which is the longest running anime

```python
# Spliting the Start Date and The End date
data[["Start","End"]]=data["Time"].str.split("-",expand=True)
data.head()
```

```python
data["Start"]=pd.to_datetime(data["Start"],format="%b %Y")
data.head()
```

```python
data["End"]=pd.to_datetime(data["End"],format="%b %Y")
data.head()
```

```python
data["Duration"]=data["End"]-data["Start"]
data.head()
```

```python
id=data["Duration"].idxmax()
print(id)
print("The Longest Runnuing Anime is :" , data.loc[id,"Title"],"\n This show has runned over :",data.loc[id,"Duration"])
```
