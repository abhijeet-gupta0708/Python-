# Missing Data

```python
data_frame={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,5],
    'c':[1,2,3,np.nan,np.nan],
    'd':[1,np.nan,np.nan,np.nan,5]
}
df=pd.DataFrame(data_frame)
df
```

Finfing Missing Element

```python
df.isna()
```

FINDING total no of missing element 

```python
df.isna().sum()
```

Removing Null value rows

```python
df.dropna()
#Since all the rows contains atleast one null values so all rows are deleted
```

```python
# Using Thresh value as COndition

df.dropna(thresh=3)
# it will return rows that contains atleast 3 known values in it
```

FILLING THE MISSING DATA

```python
df.fillna(3)
#it will replace all the Nan values to 3
```

```python
# Substituing Using a list 
values={'a':20,'b':30,'c':40,'d':50}
df.fillna(value=values)
```

```python
# Replacing All Values With their Mean
df.fillna(df.mean())
```
