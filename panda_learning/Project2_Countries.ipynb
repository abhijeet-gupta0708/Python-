```python
import pandas as pd
```

Using Link TO access the Data in place of downloading and storing data locally

```python
url="https://raw.githubusercontent.com/AkarshVyas/Pandas-Youtube/main/Countries.csv"
data=pd.read_csv(url)
data
```

```python
data.head()
```

# Q 1:-Which country has the highest population?

```python
highest=data["population"].idxmax()
print("The country with heighest Population is :",data.loc[75,"country"], "With Population of : ",data.loc[highest,"population"] ,"People")
```

# Q 2: What is the capital of the country with the highest population ?

```python
print("The Capital of COuntry WIth highest Population is : ",data.loc[highest,"capital_city"])
```

# Q 3:-  Which country has the least population?

```python
lowest=data["population"].idxmin()
print("The country with Least Population is :",data.loc[lowest,"country"], "With Population of : ",data.loc[lowest,"population"] ,"People")
```

# Q 4: - What is the capital of the country with the least population?

```python
print("The Capital of COuntry WIth lowest Population is : ",data.loc[lowest,"capital_city"])
```

 # Q 5:-Give the top 5 countries with the highest democracy score.



```python
data.sort_values("democracy_score" ,ascending = False)["country"].head(5)
```

# Q 6: - * How many total regions are there?



```python

print("There are " , data["region"].agg("sum",axis=0))
print("\n In total :",data["region"].value_counts().count(),"Region")
```

# Q 7 :-- * How many countries lie in the Eastern Europe region?


```python

print("Eastern Countries are \n: " ,data[data["region"]=="Eastern Europe"]["country"])
print("\nTotal no of Countries are : ", data[data["region"]=="Eastern Europe"]["country"].count())
```

# Q 8:- * Who is the political leader of the 2nd highest populated country?


```python
data[data['population'] == data['population'].nlargest(2).iloc[1]]['political_leader']
```

# Q 9: - * How many countries have unknown political leaders?



```python
unkonwn_leader=data[data["political_leader"].isna()]['country'].count()
print(unkonwn_leader )
```

# Q 10 :- * How many countries have “Republic” in their full name?


```python
count=0
def find_republic(txt) :
    global count
    if "republic" in txt.lower():
        count +=1
    
    return count

data["country_long"].apply(find_republic)
print(count)
```

# Q 11:- * Which country in the African continent has the highest population?


```python
african_countries=data[data["continent"]=="Africa"]['country']
africa_highest=african_countries.idxmax()
print("The Country with Highest Population in african continent is " ,data.loc[africa_highest,"country"],"With Population of ",data.loc[africa_highest,"population"],"People")
```
