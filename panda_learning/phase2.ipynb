```python
import pandas as pd
import numpy as np
```

```python
#Data Frame About Employess
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
#inner is used to connect the data of 1st frame to second frame
pd.merge(employees,salaries,on='employee_id',how='inner')
```

```python
pd.merge(employees,salaries,on='employee_id',how='outer')
```

# Using COncatination

```python
pd.concat((employees,salaries))
```

```python
# Concatination Using Axis
pd.concat((employees,salaries),axis=1)
print("\n")
pd.concat((employees,salaries),axis=0)
```

# Group By Aggregration

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data) 
df
```

```python
# Using Group Functions
df.groupby(['Category'])['Sales'].sum()
```

```python
# Using Group Functions
df.groupby(['Store'])['Quantity'].sum()
```

 Using Aggregation ,(Aggregation is used to find ['Mean','Median',Max','min','count','std','sum'])

```python
df['Sales'].agg(['sum','mean','median','max','min','std','count'])
```

# Pivot Tabels

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
obj=df.groupby(['Product','Region'])

for i,v in obj :
    print(i)
    print("\n")
    print(v)
```

```python
# Using Pivot_Table for converting and simplifiying the data
```

```python
pd.pivot_table(df,values="Sales",index="Region",columns="Product")
# it gives the mean values and gives NaN where it cannot find any data
```

```python
# CHanging the mean value to overall sum and count , max , median
pd.pivot_table(df,values="Sales",index="Region",columns="Product",aggfunc=["sum","count","max","median"])
```

CrossTabs
There is one more function which is used just as count

```python
pd.crosstab(df["Region"],df["Product"])
```
