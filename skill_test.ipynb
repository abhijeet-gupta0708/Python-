#  Q :1 :- Palindrome Number (Without Converting to String)

```python
import time
```

```python
def is_panildrome(num : int) -> bool :
    sum=0
    temp=num
    while (num !=0) :
        rem=num%10
        sum=sum*10+rem
        num //=10
    if(sum==temp) :
        return True
    else :
        return False
    
if __name__=="__main__" :
    ine=int(input("Enter a number"))
    print(is_panildrome(ine))
```

 # Q 3 : Character Frequency Counter

Write a function that:

Takes a string

Returns a dictionary with character frequencies

Ignore spaces

Case insensitive

```python
def frequency_counter(textinput:str) ->dict:
    freq={}
    text=textinput.lower()
    for  char in text :
        if char in freq :
            freq[char]+=1
        if char==" " :
            continue;
        else :
            freq[char]=1
    return freq
if __name__=="__main__":
    string=input("Enter a String ")
    result =frequency_counter(string)
    print(result)
```

# Q: 4 Find Missing Number

Given list of numbers from 1 to n with one missing, find the missing number.

```python
listt=[]
#size=int(input("Enter the value of n"))
cond=True
while(cond):
        inputt=int(input("Enter the Content of List to exit press 0"))
        listt.append(inputt)
        if inputt==0:
                cond=False

## Using Sum method to find the missing number
listt.sort()
size=listt[-1]
summ=sum(listt)
actual_sum=int((size*(size+1)//2))
missing_num=actual_sum-summ
print("Missing Number is ",missing_num)
```
