# NumPy OPerations

```python
import numpy as np
```

## SLICING THE ARRAY

```python
arr=np.arange(1,11)
print("Original Array" ,arr)
print("Basic Slicing" ,arr[2:7])
print("With steps" ,arr[1:8:2])
print("Negative Slicing" ,arr[-3])
print("Negative Slicing with End" ,arr[-3:])
print("Negative Slicing From Start" ,arr[:-3])
```

# Slice 2 D or n D array

```python
arr2d=np.array([[3,1],[1,2],[2,3]])
print("Sorted Array By Column : ",np.sort(arr2d,axis=0))
print("Sorted Array By Row : ",np.sort(arr2d,axis=1))
```

# Filters

```python
num=np.arange(1,11)
even=num[num%2==0]
print("Even Nos are : ",even)

# MASk (mask is the same expression as  FIller)

mask=(num%2)
print("Even Nos are Using Mask : ",num[mask==0])
mask=num>5
print("Num greater than '5' ",num[mask])
```

# np.where()

```python
# Using Array from Above cell
indices=[0,1,4]
print(num[indices])
where_result=np.where(num>5)
print("Np where is used " ,num[where_result])
```

## ADD OR REMOVE ELEMENTS

```python
arr=np.arange(1,5)
arr2=np.arange(4,8)
 # Performing Normal Addition 
add=arr+arr2
print("Addition of Two Arrays : ",add)

# Additing the elements into a single array 

conc=np.concatenate((arr,arr2))
print("COncated Array is : " ,conc)
```

# Adding Row And COlumn

```python
original=np.array([[1,2],[3,4],[5,6]])
new=np.array([5,6])
newC=np.array([[7],[8],[9]])
#Adding ROw using np.vstack
rowarr=np.vstack((original,new))
#Adding Column using np.hstack
coluarr=np.hstack((original,newC))
print("Original Array : ",original)
print("Row is added in Array : ",rowarr)
print("Column is added Array : ",coluarr)
```

# delete

```python
delete=np.delete(original,2)
print("Array after deletion : ",delete)
```
